1. Explain tokens, context windows, inference, and hallucinations

Solution: Tokens: The fundamental chunks or sub-word fragments of text that language models process. A token can be a full word, part of a word (like a prefix or suffix), a single character, or punctuation. On average, 1 token is roughly equivalent to 3/4 of an English word (or 100 tokens being around 75 words).

Context Windows: The maximum limit of tokens (both input prompt and generated output combined) that a language model can process and retain in active memory at any single time. If the total token length exceeds this window, older context is dropped or truncated.

Inference: The runtime process of a trained language model taking an input prompt, passing it through its neural network parameters and attention layers, and generating new token predictions or outputs.

Hallucinations: Confident outputs generated by a language model that are factually incorrect, fabricated, or unsupported by the training data and context provided, often presented with misleading certainty.

2. Write prompts for summarization, extraction, classification, and question answering.


Solution: Summarization Prompt:

"Read the following quarterly financial report and synthesize the core points into exactly three concise bullet points focusing solely on total revenue, net profit margins, and next-quarter guidance. Exclude all filler text and background narrative: [Insert Text]"

Data Extraction Prompt:

"Analyze the unstructured vendor invoice text below. Extract all entity values into a valid, strict JSON object containing the following keys: vendor_name, invoice_date, line_items (as an array of objects containing description and price), and total_amountDue: [Insert Text]"

Classification Prompt:

"Classify the sentiment of the following customer service support ticket into one of three strict categories: Positive, Neutral, or Negative. Output your response as a JSON dictionary containing the category and a single-sentence justification: [Insert Text]"

Question Answering Prompt:

"Using strictly and only the provided reference context below, answer the user's explicit question. If the correct answer cannot be found or deduced directly from the text, output exactly: 'I cannot answer this based on the provided context.' Do not use outside knowledge.
Context: [Insert Context]
Question: [Insert Question]"

3. Compare basic prompts with improved prompts on ten examples.


Solution: Here are 10 distinct test examples showing a basic prompt versus an improved prompt designed to enforce constraints, context, and structural formatting.

Summarization

Input Text: A 2-page tech blog post detailing a company's cloud infrastructure migration from AWS to Google Cloud Platform, noting cost reductions of 20%, downtime of 4 hours during the cutover, and team training hurdles.

Basic Prompt: "Summarize this."

Improved Prompt: "Summarize the following cloud migration text into exactly two bullet points focusing strictly on financial savings and technical downtime. Omit background narratives."

Data Extraction

Input Text: "Order ID: ORD-9921, Customer: John Smith, Email: john@example.com, Total Amount: $145.50, Status: Shipped."

Basic Prompt: "Extract the customer and order info."

Improved Prompt: "Extract the order details from the text below and output them as a strict JSON object with keys: order_id, customer_name, email, total_amount, and status."

Classification

Input Text: "The application interface looks clean, but the database connection keeps timing out during peak hours, causing data loss."

Basic Prompt: "What kind of review is this?"

Improved Prompt: "Classify the sentiment of the following feedback into one of three categories (Positive, Neutral, Negative) and provide a one-sentence technical justification."

Question Answering

Input Text: "SentinelPay is a sub-100ms real-time payment risk engine built with Java Spring Boot and PostgreSQL, running inside Docker on Render."

Basic Prompt: "What database does SentinelPay use?"

Improved Prompt: "Using strictly the provided text, answer what database SentinelPay uses. If it is not mentioned, state 'Not specified'."

Code Generation

Input Text: Write a function to check if a string is a palindrome.

Basic Prompt: "Write a Python function for a palindrome."

Improved Prompt: "Write a production-ready Python function to check if a given string is a palindrome. Include type hinting, docstrings, and handle case-insensitivity and whitespace stripping."

Text Transformation / Rewriting

Input Text: "Hey boss, the server crashed because someone deployed bad code without testing. We fixed it though."

Basic Prompt: "Rewrite this professionally."

Improved Prompt: "Rewrite the following informal message into a formal incident report update for executive stakeholders, maintaining a professional and objective tone."

Entity Recognition

Input Text: "Dr. Alan Turing attended the Association for Computing Machinery conference in New York on May 12, 1952."

Basic Prompt: "Find the entities in this sentence."

Improved Prompt: "Extract all Person names, Organizations, Locations, and Dates from the text below into a structured YAML format."

Logical Reasoning (Math/Word Problem)

Input Text: "A data pipeline processes 120 records per minute for 1 hour, then slows down by 25% for the next hour."

Basic Prompt: "How many records are processed?"

Improved Prompt: "Solve the word problem step-by-step. Show your intermediate arithmetic reasoning inside <scratchpad> tags before stating the final total numerical answer."

Debugging Code

Input Text: print(x) where x is undefined, throwing a NameError.

Basic Prompt: "Fix this code error."

Improved Prompt: "Analyze the following Python error traceback (NameError). Explain the root cause in one sentence and provide the corrected code snippet."

Creative Writing / Marketing

Input Text: A cybersecurity tool that blocks zero-day exploits.

Basic Prompt: "Write an ad for this security tool."

Improved Prompt: "Write a high-converting, punchy LinkedIn advertisement copy (max 40 words) for an enterprise cybersecurity tool targeting Chief Information Security Officers (CISOs). Include a clear call to action."

4. Record correctness and instruction-following errors.


Solution: While running and testing prompts, errors typically fall into two categories:

Correctness Errors (Factual Inaccuracies / Hallucinations):

Description: The model generates statements that contradict the source text, introduces false external facts, or calculates incorrect mathematical outputs.

Example observed in testing: When using basic prompts for question answering, the model sometimes hallucinates architectural specs not present in the text (e.g., claiming a system uses Redis when only PostgreSQL was mentioned).

Instruction-Following Errors (Constraint Violations):

Description: The model fails to adhere to structural, formatting, or negative constraints specified in the prompt (e.g., outputting paragraphs instead of requested JSON, exceeding word limits, or ignoring persona instructions).

Example observed in testing: A basic prompt asking for "exactly two bullet points" returns a conversational paragraph with four bullet points instead. Improved prompts using strict delimiter tags and JSON schemas drastically minimize this error.

5. Explain the difference between prompting, RAG, and fine-tuning.


Solution: Prompting:

Technical & Operational nature: Relies entirely on the base pre-trained model's frozen weights without modifying any internal parameters. Customization is achieved dynamically at inference time by crafting precise instructions, context windows, few-shot examples, and system personas. It is fast, requires zero training data, but is limited by the model's context length and susceptibility to hallucinations.

Retrieval-Augmented Generation (RAG):

Technical & Operational nature: A hybrid architecture that connects the LLM to an external knowledge base or vector database. When a query is made, relevant documents are retrieved via semantic search and injected dynamically into the prompt's context window. This grounds the model's responses in factual, verifiable external data without retraining the neural network.

Fine-tuning:

Technical & Operational nature: An active training process where the model's internal weights and biases are updated by training on a curated dataset of domain-specific input-output pairs. This alters the model's permanent behavior, tone, or style, allowing it to internalize complex jargon and specialized tasks natively without needing massive prompt context.
